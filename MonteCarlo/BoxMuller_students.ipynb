# Generating Gaussian random numbers with the Box–Muller transform

The **Box–Muller transform** turns pairs of *uniform* random numbers into pairs of *independent, standard normal* random numbers ($\mu=0$, $\sigma=1$).

**Algorithm**

1. Draw two independent uniform numbers $u_1, u_2$ in $(0,1]$.
2. Set
   $$\phi = 2\pi u_1, \qquad r = \sqrt{-2\ln u_2}.$$
3. Then
   $$z_1 = r\cos\phi, \qquad z_2 = r\sin\phi$$
   are two independent random variables, each following $N(0,1)$.

**Why it works (sketch).** If $u_2 = e^{-r^2/2}$ is uniform, then $\frac{dp}{dr} = \left|\frac{du_2}{dr}\right| = r\,e^{-r^2/2}$. Together with a uniform $\phi$ this gives
$$dp = \frac{1}{2\pi}e^{-r^2/2}\,r\,dr\,d\phi = \frac{1}{2\pi}e^{-(z_1^2+z_2^2)/2}\,dz_1\,dz_2 ,$$
i.e. the 2D standard normal density, since $r\,dr\,d\phi = dz_1\,dz_2$.

> **Note.** `rng.random()` returns values in $[0,1)$, so $\ln u_2$ could diverge for $u_2=0$. We use $1-u$ instead, which lies in $(0,1]$.
